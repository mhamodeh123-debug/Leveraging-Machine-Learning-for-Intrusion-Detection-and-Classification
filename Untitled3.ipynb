```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier, AdaBoostClassifier, StackingClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import seaborn as sns
import time
```

```python
data = pd.read_csv("train_test_network.csv")
np.random.seed(0)
data.head()
```

```python
data.info()
```

```python
data.columns
```

```python
df = pd.DataFrame(data)

null_counts = df.isnull().sum()
print(null_counts)

rows_with_missing = df.isnull().any(axis=1).sum()
print("Total rows with at least one missing value:", rows_with_missing)
```

```python
# Drop duplicates
df.drop_duplicates(inplace=True)

# Fill missing values with median
df.fillna(df.median(numeric_only=True), inplace=True)
```

```python
label_col = 'label'   # change if your label column name is different

if df[label_col].dtype == 'object':
    encoder = LabelEncoder()
    df[label_col] = encoder.fit_transform(df[label_col])
```

```python
X = df.drop(label_col, axis=1)
y = df[label_col]
```

```python
# Drop IP address columns (avoid data leakage)
ip_cols = [col for col in X.columns if 'ip' in col.lower()]
print("Dropped IP columns:", ip_cols)

X.drop(columns=ip_cols, inplace=True)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

```

```python
# Detect categorical (non-numeric) columns
categorical_cols = X.select_dtypes(include=['object']).columns
print("Categorical columns:", categorical_cols)

# Encode categorical columns using LabelEncoder
for col in categorical_cols:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
def evaluate_model(model, X_train, X_test, y_train, y_test):
    start = time.time()
    model.fit(X_train, y_train)
    end = time.time()
    
    y_pred = model.predict(X_test)
    
    return {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "Inference Time": end - start
    }
```

```python
models = {
    "Naive Bayes": GaussianNB(),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(n_estimators=100),
    "SVM": SVC(kernel='rbf'),
    "Bagging": BaggingClassifier(
        estimator=DecisionTreeClassifier(),
        n_estimators=50
    ),
    "AdaBoost": AdaBoostClassifier(n_estimators=50)
}
```

```python
results = {}

for name, model in models.items():
    results[name] = evaluate_model(
        model, X_train, X_test, y_train, y_test
    )
```

```python
results_df = pd.DataFrame(results).T
results_df
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=results_df.index, y=results_df["Accuracy"])
plt.xticks(rotation=45)
plt.title("Model Accuracy Comparison")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(results_df.iloc[:, :-1], annot=True, fmt=".2f", cmap="viridis")
plt.title("Model Performance Metrics")
plt.show()
```
